```python
##import liberaries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
##read csv file
df=pd.read_csv("data_YesBank_StockPrices.csv")
```

```python
df.head()
```

## Data cleaning step and Eda

```python
## data cleaning step
df.isnull().sum()
```

```python
df.info
```

```python
##convert date
df["Date"]=pd.to_datetime(df["Date"],format="%b-%y")
```

```python
##Extract month and year
df["Month"]=df["Date"].dt.month
df["Year"]=df["Date"].dt.year
```

```python
df["Date"]=df["Date"].map(pd.Timestamp.toordinal)
```

```python
df.head()
```

## data visulazation

```python
## data visulazation
sns.pairplot(df)
```

```python
## dependent and independent variable
x=df.drop("Close",axis=1)
y=df["Close"]
```

```python
x
```

## train_test_split process

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.25,random_state=42)
```

```python
x_train.corr()
```

```python
##multicollinarity check
plt.figure(figsize=(12,10))
corr=x_train.corr()
sns.heatmap(corr,annot=True)
plt.show()
```

## import linear regression and model training

```python
from sklearn.linear_model import LinearRegression
model=LinearRegression()
```

```python
model.fit(x_train,y_train)
```

```python
print(df.dtypes)
```

```python
y_pred=model.predict(x_test)
```

```python
##Accuracy
from sklearn.metrics import r2_score,mean_ab
score=r2_score(y_test,y_pred)
```

```python
print("Accuracy",score)
```

```python
from sklearn.metrics import mean_absolute_error
from sklearn.metrics import r2_score
me=mean_absolute_error(y_test,y_pred)
score=r2_score(y_test,y_pred)
print("mean absolute error",me)
print("R2 score",score)
```

```python
plt.scatter(y_test,y_pred,color='blue')
plt.show()
```

```python
train_r3=model.score(x_train,y_train)
```

```python
##model checking strong or not
test_r3=model.score(x_test,y_test)
print("Train_r3",train_r3)
print("test_r3",test_r3)
```
