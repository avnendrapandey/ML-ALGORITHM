## data cleaning and EDA (exploratorary analysis )

```python
df['Amount']=df['Quantity']*df['UnitPrice']
```

```python
df['InvoiceDate']=pd.to_datetime(df['InvoiceDate'])
```

```python
##latest data in dataset
snap_shot_date=df['InvoiceDate'].max()+pd.Timedelta(days=1)
```

```python
## RFM table
RFM=df.groupby('CustomerID').agg({
    'InvoiceDate':lambda x:(snap_shot_date-x.max()).days,
    'InvoiceNo':'nunique',
    'Amount':'sum'
})
RFM.columns=['Recency','Frequency','Monetary']
```

```python
print(RFM.head())
```

## standardization

```python
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
rfm_scaled=scaler.fit_transform(RFM)
```

```python
rfm_scaled=pd.DataFrame(
    rfm_scaled,
    columns=RFM.columns,
    index=RFM.index
)
```

```python
from sklearn.preprocessing import OneHotEncoder,LabelEncoder
```

```python
ohe=OneHotEncoder()
lb=LabelEncoder()
```

```python
df['InvoiceNo']=lb.fit_transform(df['InvoiceNo'])
df['StockCode']=lb.fit_transform(df['StockCode'])
df['Description']=lb.fit_transform(df['Description'])
```

```python
df['Country']=lb.fit_transform(df['Country'])
```

```python
df['UnitPrice']=df['UnitPrice'].astype(int)
df['CustomerID']=df['CustomerID'].fillna(0).astype(int)
df['Amount']=df['Amount'].fillna(0).astype(int)
```

```python
df.head()
```

## visulazation 

```python
df['Sales']=df['Quantity']*df['UnitPrice']
country_Sales=df.groupby('Country')['Sales'].sum().sort_values(ascending=False).head(10)
country_Sales.plot(kind='bar')
plt.title('Top 10 countries by sales')
plt.xlabel('Country')
plt.ylabel('Sales')
plt.show()
```

```python
## monthly sales trend
df['InvoiceDate']=pd.to_datetime(df['InvoiceDate'])
monthly_sales=df.groupby(df['InvoiceDate'].dt.to_period('M'))['Sales'].sum()
monthly_sales.plot(kind='line')
plt.title('Monthly Sales trend')
plt.xlabel('Month')
plt.ylabel('Sales')
plt.show()
```

```python
## top 10 product sold
top_products=df.groupby('Description')['Quantity'].sum().sort_values(ascending=False).head(10)
top_products.plot(kind='bar')
plt.title('Top 10 product sales')
plt.xlabel('Product')
plt.ylabel('Quantity sold')
plt.show()
```

```python
## top customers by revenue
top_customers=df.groupby('CustomerID')['Sales'].sum().sort_values(ascending=False).head(10)
top_customers.plot(kind='bar')
plt.title('Top 10 customer')
plt.xlabel('customerid')
plt.ylabel('Revenue')
plt.show()
```

```python
## Elbow Method
from sklearn.cluster import KMeans
wcss=[]

for k in range(2,11):
    kmeans=KMeans(n_clusters=k,random_state=42,n_init=10)
    kmeans.fit(rfm_scaled)
    wcss.append(kmeans.inertia_)

plt.plot(range(2,11),wcss,marker='o')
plt.xlabel('Number of clusters')
plt.ylabel('wcss')
plt.title('Elbow Method')
plt.show()
```

```python
## silhoutte score
from sklearn.metrics import silhouette_score

for k in range(2,11):
    model=KMeans(n_clusters=k,random_state=42,n_init=10)
    labels=model.fit_predict(rfm_scaled)
    score=silhouette_score(rfm_scaled,labels)
    print(f'k={k}-->{score:3f}')
    
```

```python
best_k=2
k_means=KMeans(n_clusters=best_k,random_state=42,n_init=10)
RFM['cluster']=kmeans.fit_predict(rfm_scaled)
```

```python
cluster_summary=RFM.groupby('cluster').mean()
print(cluster_summary)
```

```python
segment_map={
    0:'High Value',
    1:'Regular',
    2:'At Risk',
    3:'Ocassional'
}
RFM['Segment']=RFM['cluster'].map(segment_map)
```

```python
##2d visulazation
plt.figure(figsize=(8,6))
plt.scatter(
    RFM['Frequency'],
    RFM['Monetary'],
    c=RFM['cluster']
)
plt.xlabel('Frequency')
plt.ylabel('Monetary')
plt.title('customer segments')
plt.show()
```

```python
## 3d visulazation
from mpl_toolkits.mplot3d import Axes3D
fig=plt.figure(figsize=(8,6))
ax=fig.add_subplot(111,projection='3d')
ax.scatter(
    RFM['Recency'],
    RFM['Frequency'],
    RFM['Monetary'],
    c=RFM['cluster']
)
ax.set_xlabel('Recency')
ax.set_ylabel('Frequency')
ax.set_zlabel('Monetary')
plt.show()
```

```python
from sklearn.metrics import calinski_harabasz_score
score=calinski_harabasz_score(
    rfm_scaled,
    RFM['cluster']
)
print("ch score",score)
```

```python
##customer cluster merge
customer_cluster=RFM[['cluster']].reset_index()
df_cluster=df.merge(
    customer_cluster,
    on='CustomerID',
    how='left'
)
#top 5 recommendation
recommendations={}
for cluster in df_cluster['cluster'].unique():
    top_products=(
        df_cluster[df_cluster['cluster']==cluster].groupby('Description')['Quantity'].sum().sort_values(ascending=False).head(5).index.tolist())
    recommendations[cluster]=top_products
```

```python
import joblib
joblib.dump(recommendations,'recommendations.pkl')
```

```python
%%writefile app.py
import streamlit as st
import joblib
model=joblib.load('customer_segementation.pkl')
scaler=joblib.load('scaler.pkl')
recommendations=joblib.load('recommendations.pkl')
st.title('customer segmentation and recommendation')
recency=st.number_input('Recency')
frequency=st.number_input('Frequency')
monetary=st.number_input('Monetary')

if st.button('predict segment'):
    x=scaler.transform(
        [[recency,frequency,monetary]]
    )
    cluster=model.predict(x)[0]
    st.success(f"cluster:{cluster}")
    st.subheader("Recommended products")
    for p in recommendations[cluster]:
        st.write("Right",p)
```
